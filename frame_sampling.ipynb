set up output and run script

```python
video_directory = "drive/MyDrive/maple_resource/test_video.mp4" # @param {"type":"string","placeholder":"video directory"}
# @markdown video directory
# @markdown
# @markdown it should be the same video file as above
output_csv = "drive/MyDrive/maple_resource/result-mulung.csv" # @param {"type":"string","placeholder":"output directory"}
# @markdown the output csv file
debug_dir = "drive/MyDrive/maple_resource/debug_crops" # @param {"type":"string","placeholder":"debug directory"}
# @markdown the debug screenshot directory
mask_dir = "drive/MyDrive/maple_resource/masks" # @param {"type":"string","placeholder":"mask directory"}
# @markdown the directory that stores the masks
buff_dir = "drive/MyDrive/maple_resource/buff_icons"# @param {"type":"string","placeholder":"buff_icons"}
# @markdown the directory that stores the buff icons
debuff_dir = "drive/MyDrive/maple_resource/debuff_icons"# @param {"type":"string","placeholder":"debuff icons"}
# @markdown the directory that stores the debugg  icons
start_time = "8m40s" # @param {"type":"string","placeholder":"video timestamp"}
end_time = "13m40s" # @param {"type":"string","placeholder":"video timestamp"}
interval = 1 # @param {"type":"slider","min":0,"max":10,"step":1}
```

### Some basic checks to ensure that all files are accounted

```python

print("Video directory:",Path(video_directory).is_file())
print("Masks:",[x.name for x in Path(mask_dir).iterdir()])
print("Buffs:",[x.name for x in Path(buff_dir).iterdir()])
print("Buffs:",[x.name for x in Path(debuff_dir).iterdir()])
```

```python
import subprocess

process = subprocess.run(["python", "ba_extractor.py",
                "--file",video_directory,
                "--debug_dir",debug_dir,
                "--output",output_csv,
                "--mask_dir",mask_dir,
                "--buff_dir",buff_dir,
                "--debuff_dir",debuff_dir,
                "--start_time",start_time,
                "--end_time",end_time,
                "--interval",str(interval)],
               text=True)

# The return code can still be accessed after the process completes
print("Return Code:", process.returncode)
```
