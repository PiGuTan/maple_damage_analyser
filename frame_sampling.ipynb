```python
from objects import Video
from util import time_to_seconds
import cv2
import opencv_jupyter_ui as jcv2
```

#### edit video path and timestamp
there will be an output showing the start_frame, end_frame and approximate frame rate
see if they make sense

```python
video = Video("test_videos/carling_high_res.mp4")
print("All Frames:",video.frames)
time_in_seconds:int = time_to_seconds("3m02s")
```

```python
frame_index = video.get_frame_index_with_time_seconds(time_in_seconds)
print("Frame Index:",frame_index)
print("Frame time",f"{video.get_frame_time(frame_index):.3f}s")
```

#### Click this to go to next frame

```python
frame_index +=1
print("Frame Index:",frame_index)
print("Frame time",f"{video.get_frame_time(frame_index):.3f}s")
```

#### Click this to go to previous frame

```python
frame_index -=1
print("Frame Index:",frame_index)
print("Frame time",f"{video.get_frame_time(frame_index):.3f}s")
```

### Check if frame have what you want
if it doesnt, you can:
1. go to next frame
2. go to previous frame
3. readjust timestamp

```python
output_frame = video.get_frame(frame_index)
jcv2.imshow(f"Frame Display {frame_index}", output_frame)
```

## Happy with the frame? Save it

```python
frame_time = f"{video.get_frame_time(frame_index):.3f}"
cv2.imwrite(f"extracted_frame_{frame_time}.png", output_frame)
```
